```python
import os
import matplotlib.pyplot as plt
import torch
from PIL import Image
from torchvision import transforms
from tqdm import tqdm
```

```python
transform = transforms.ToTensor()
```

```python
def get_cloud_masks(split="train"):

    dico = {}

    cpt = 0
    
    for (root, dirs, files) in os.walk(f"wildfires-cems/{split}", topdown=True):
        for file in files:
            if "_CM" in file and (file.endswith(".png")):

                cpt += 1
                print(cpt)
                if cpt > 10:
                    return dico

                path_to_img = os.path.join(root, file)
                img = Image.open(path_to_img).convert("RGB")
                tensor = (transform(img) * 255).byte().reshape(3, -1)

                pixels_values = torch.unique(tensor, dim=1, return_counts=True)

                nb_classes = pixels_values[1].shape[0]

                classes = []
                for classe in range(nb_classes):
                    valeur_pixel = pixels_values[0][:, classe]
                    nb_pixels = pixels_values[1][classe]
                    if torch.equal(valeur_pixel, torch.tensor([103, 190, 224], dtype=torch.uint8)):
                        classes.append(f"Clear : {nb_pixels.item()}")
                    elif torch.equal(valeur_pixel, torch.tensor([180, 180, 180], dtype=torch.uint8)):
                        classes.append(f"Clouds : {nb_pixels.item()}")
                    elif torch.equal(valeur_pixel, torch.tensor([220, 220, 220], dtype=torch.uint8)):
                        classes.append(f"Light clouds : {nb_pixels.item()}")
                    elif torch.equal(valeur_pixel, torch.tensor([60, 60, 60], dtype=torch.uint8)):
                        classes.append(f"Shadow : {nb_pixels.item()}")
                
                dico[path_to_img] = classes

    return dico

dictionnaire1 = get_cloud_masks("train")
print(dictionnaire1)
```

```python
def get_wildfire_masks(split="train"):

    dico = {}

    cpt = 0
    
    for (root, dirs, files) in os.walk(f"wildfires-cems/{split}", topdown=True):
        for file in files:
            if "_GRA" in file and (file.endswith(".png")):

                cpt += 1
                print(cpt)
                if cpt > 10:
                    return dico

                path_to_img = os.path.join(root, file)
                img = Image.open(path_to_img).convert("RGB")
                tensor = (transform(img) * 255).byte().reshape(3, -1)

                pixels_values = torch.unique(tensor, dim=1, return_counts=True)

                nb_classes = pixels_values[1].shape[0]

                classes = []
                for classe in range(nb_classes):
                    valeur_pixel = pixels_values[0][:, classe]
                    nb_pixels = pixels_values[1][classe]
                    if torch.equal(valeur_pixel, torch.tensor([0, 0, 0], dtype=torch.uint8)):
                        classes.append(f"No visible damage : {nb_pixels.item()}")
                    elif torch.equal(valeur_pixel, torch.tensor([254, 217, 142], dtype=torch.uint8)):
                        classes.append(f"Moderate damage : {nb_pixels.item()}")
                    elif torch.equal(valeur_pixel, torch.tensor([254, 153, 41], dtype=torch.uint8)):
                        classes.append(f"High damage : {nb_pixels.item()}")
                    elif torch.equal(valeur_pixel, torch.tensor([204, 76, 2], dtype=torch.uint8)):
                        classes.append(f"Destruction : {nb_pixels.item()}")
                    
                dico[path_to_img] = classes

    return dico

dictionnaire2 = get_wildfire_masks("train")
print(dictionnaire2)
```
